## File Handeling
r->read,w->write,a->append

```python
## Reading a file
file=open("untitled.txt","r")
content=file.read()
print(content)
file.close()
```

```python
## Reading a file (first line)
file=open("untitled.txt","r")
content=file.readline()
print(content)
file.close()
```

```python
## Reading a file (printing all points in one line)
file=open("untitled.txt","r")
content=file.readlines()
print(content)
file.close()
```

```python
## writing a file
file=open("untitled_01","w")
content=file.write("Welcome Here")
file.close()

file=open("untitled_01","r")
content=file.read()
print(content)
file.close()
```

```python
## Adding Data to File
file=open("untitled_01","a")
content=file.write("\nThis is Computer Lab")
file.close

file=open("untitled_01","r")
content=file.read()
print(content)
file.close
```
