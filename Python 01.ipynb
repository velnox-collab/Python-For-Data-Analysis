## Variables:
* Allowed Characters: Variable names can only contain letters (A-Z, a-z), digits (0-9), and underscores (_). No spaces or special characters (like !, @, #, $, %) are allowed.

* Starting Character: A variable must start with a letter or an underscore. It cannot start with a number.• _user_id and user1 are valid.
1user is invalid.

* Case-Sensitivity: Python treats lowercase and uppercase letters differently. age, Age, and AGE are treated as three separate variables.

* Reserved Keywords: You cannot use Python's built-in keywords (such as if, else, while, for, class, or def) as variable names because they are reserved for the language's core functionality

## Data Structures

* List: A dynamic array used to store sequence collections.
    

```python
names=["ali","ahmed","sara"]
print(names)
print(type(names))
```

```python
names.append("naeem") # adds value at the end of the list
print(names)
```

```python
names.count("ali") # Counts how much time this value is written inside an list
```

```python
names.index("sara") # finds the index of a specific value
```

```python
names.insert(1,"afzal") # insert value at specific index
print(names)
```

```python
names.pop() # returns and removes any 1 value from the list
```

```python
names.remove("afzal") 
print(names)
```

```python
list1=[2,7,4,2,0,9,7]
list1
```

```python
list1.sort() # Sorts the values
print(list1)
```
