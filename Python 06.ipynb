## Advance Python

```python
# Decorators: It decorates our functions...like an topping on a cake

def School(func):
    def wrapper():
        print("Student name:")
        func()
        print("He is above average student ")
    return wrapper

@School
def Student():
    print("Bashir")

Student()
```

```python
def total(func):
    def wrapper(a,b):
        print("your total is")
        func(a,b)
        print("Answer")
    return wrapper

@total
def addition(a,b):
    print(f"Total {a+b}")

addition(9,4)
```

```python
## *args: it stores multiple arguments in the form of a tuple
def total(*args):
    sum = 0
    for i in args:
        sum=sum+i

    print(sum)
print(total(5,5))
print(total(3,4))
```

```python
## **kwargs:It creates dictionary

def data(**kwargs):
    print(kwargs)

data(a=5,b=9,c=1)
    
```

```python
## *args + **kwargs + decorators

def total(func):
    def wrapper(*a,**b):
        print("your total is")
        func(*a,**b)
        print("Answer")
    return wrapper

@total
def addition(a,b,c,d):
    print(a+b+c+d)

addition(45,6,7,9)
```

```python
## list Comprehension

l=[i for i in range(1,21) if i % 2==0]
print(l)

## Dictionary Comprehension

l={i:i**2 for i in range(1,10)}
print(l)

## Set Comprehension

l={i for i in range(1,11) if i % 2==0}
print(l)
```

```python
## Lambda function

addition=lambda a,b:a + b
print(addition(8,9))

even_odd = lambda a: "even" if a % 2 ==0 else "odd"
print(even_odd(9))
```
