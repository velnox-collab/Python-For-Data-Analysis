```python
class Student:
    def __init__(self,name,age):
        self.name=name
        self.age=age

student2=Student("Ali",20)
print(student2.name)
print(student2.age)
```

```python
## Methods: Functions inside class
class Male:
    def __init__(self,name,height):
        self.name=name
        self.height=height
    def result(self):
        print(self.name,"is",self.height,"ft tall")

Man=Male("Ahmed",5.4)
print(Man.name," ",Man.height)
print(Man.result())
```

##Pillars of Oops
* Encapsulation
* Inheritance
* Polymorphism
* Abstraction

```python
### 1.Encapsulation: Data + methods ko eik unit(class) mein bund karna or data ko controlled access dena.

class BankAccount:
    def __init__(self,balance):
        self.__balance = balance

    def deposit(self,amount):
        self.__balance += amount

    def get_balance(self):
        return self.__balance
```

```python
### 2.Inheritance:Eik class doosri class ki properties aur methods ko inherit kar leti hai....

class animal:
    def eat(self):
        print("Eating")

class Dog(animal):
    def bark(self):
        print("barking")

dog=Dog()
dog.eat()
dog.bark()
```

```python
### 3.Polymorphism:same interface/method different behaviour..
### Poly=many,morph=forms

class Dog:
    def sound(self):
        print("Bark")

class cat:
    def sound(self):
        print("meow")
Dog=Dog()
cat=cat()
Dog.sound()
cat.sound()
```

```python
## 4.Abstraction:Complex implementation ko hide karna aur user ko
## sirf necessary interface dena...

class car:
    def start(self):
        self.__check_engine()
        self.__inject_fuel()
        print("car started")

    def __check_engine(self):
        pass

    def __inject_fuel(self):
        pass

car=car()
car.start()
        
```
