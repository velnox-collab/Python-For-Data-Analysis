```python
## Arithmatic Operators
a = 15
b = 4

print("Addition:", a + b)  

print("Subtraction:", a - b) 

print("Multiplication:", a * b)  

print("Division:", a / b) 

print("Floor Division:", a // b)  

print("Modulus:", a % b) 

print("Exponentiation:", a ** b)
```

```python
## Comparison Operators

a = 13
b = 33

print(a > b)
print(a < b)
print(a == b)
print(a != b)
print(a >= b)
print(a <= b)
```

```python
## Logical Operator

a = True
b = False
print(a and b)
print(a or b)
print(not a)
```

```python
## Assignment Operators

a = 10
b = a
print(b)
b += a
print(b)
b -= a
print(b)
b *= a
print(b)
b <<= a
print(b)
```

```python
## Identity Operators

a = 10
b = 20
c = a

print(a is not b)
print(a is c)
```

```python
## Membership Operators

x = 24
y = 20
my_list = [10, 20, 30, 40, 50]

if (x not in my_list):
    print("x is NOT present in given list")
else:
    print("x is present in given list")

if (y in my_list):
    print("y is present in given list")
else:
    print("y is NOT present in given list")
```

* Loops

```python
## Finding total sum using for loops
numbers = [10, 20, 5, 15]
sum=0
for i in numbers:
    sum+=i

print(sum)
```

```python
## finding even numbers using for loops
numbers = [3, 8, 11, 14, 17, 20]
for i in numbers:
    if i % 2 == 0:
        print(i)
```

```python
## finding how many numbers were greater than 50 in the given list
marks = [45, 67, 32, 89, 51, 40, 76]
for i in marks:
    if i > 50:
        print(i)
```

* Logic Building with for loop

```python
## Finding largest number without using max
numbers = [12, 45, 7, 89, 34, 23]
total=0
for i in numbers:
    total+=i*2
    
print(total)
        
```

```python
## Finding largest number without using max
numbers = [12, 45, 7, 89, 34, 23]
total=0
for i in numbers:
    total+=i*2
    maxm=total/len(numbers)
    if i > maxm:
        print(i)
    
        
```
